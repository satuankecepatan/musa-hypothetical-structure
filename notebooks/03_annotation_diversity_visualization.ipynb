## 1. Scientific Overview & Deliverable Scope
Once the ancestral *Musa* plastome sequence is synthesized, we must annotate and validate its biological functionality:
- **Coordinate Lift-Over:** Transfer canonical CDS, tRNA, and rRNA features from curated *Musa acuminata* reference (HF677508).
- **Trans-spliced $rps12$ Modeling:** Explicitly resolve the bipartite structure of $rps12$ (Exon 1 in LSC, Exons 2–3 in IRs).
- **Boundary Pseudogene Delimitation ($\psi ycf1$):** Properly flag the truncated boundary copy at the $J_{SA}$ junction as a pseudogene.
- **NCBI Table 11 Translation Audit:** Translate every coding sequence using the plastid genetic code, ensuring zero premature stop codons and accounting for RNA-edited start codons ($ndhD$ $\text{ACG} \to \text{AUG}$).
- **Sliding-Window Diversity Landscape:** Projecting modern *Musa* accessions onto the ancestral backbone, computing nucleotide diversity ($\pi$), Shannon entropy ($H$), and GC skew.
- **Publication-Grade Visualizations:** Producing 600 DPI publication figures adhering strictly to Edward Tufte minimalist principles (pure white canvas, zero plot titles, external legends).
